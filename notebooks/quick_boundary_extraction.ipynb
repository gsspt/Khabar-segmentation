```python
import json
from transformers import AutoTokenizer

# Charger les données
tokenizer = AutoTokenizer.from_pretrained("CAMeL-Lab/bert-base-arabic-camelbert-msa")
with open('data/processed/kitab_uqala_reference_corpus.txt', 'r', encoding='utf-8') as f:
    text = f.read()
with open('results/camelbert_kitab_uqala_raw_inference.json', 'r', encoding='utf-8') as f:
    predictions = json.load(f)['inference_results']['predictions']

# Tokenizer et extraire les boundary tokens
tokens = tokenizer.tokenize(text)
boundary_tokens = [t for t, p in zip(tokens, predictions[:len(tokens)]) if p == 1]
boundary_indices = [i for i, p in enumerate(predictions[:len(tokens)]) if p == 1]

# Sauvegarder
results = {
    'metadata': {
        'corpus': 'kitab_uqala_reference_corpus.txt',
        'model': 'CAMeL-Lab/bert-base-arabic-camelbert-msa',
        'total_boundary_tokens': len(boundary_tokens),
        'boundary_percentage': round(100 * len(boundary_tokens) / len(tokens), 2),
    },
    'boundary_tokens': boundary_tokens,
    'boundary_indices': boundary_indices,
}

with open('results/camelbert_boundary_tokens_clean.json', 'w', encoding='utf-8') as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print(f"✓ Done!")
print(f"  Boundary tokens: {len(boundary_tokens):,}")
print(f"  Percentage: {results['metadata']['boundary_percentage']}%")
print(f"  Saved to: results/camelbert_boundary_tokens_clean.json")
```
